IMPORTS

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler

import os
```

FILE_READ

```python
def file_read(path): # Reading the CSV file and assigning input and output
    df =pd.read_csv(path)
    X = df.iloc[:, 0:13].values
    Y = df.iloc[: , 13].values
    return X,Y
```

```python
path = 'c:/Users/haZAR/Desktop/heart_disease_data.csv'
X,Y = file_read(path)
```

TRANSFORMATIONS

```python
# Initialize the StandardScaler
ss = StandardScaler()

# Fit to the data and then transform it
# StandardScaler standardizes features by removing the mean and scaling to unit variance
# This is often a good practice, especially for algorithms such as KNN, which are distance based and require scaled data.
X = ss.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=32)
# Spliting the dataset into train_test
```

MODEL

```python

# Initialize the KNeighborsClassifier with 4 neighbors
knn = KNeighborsClassifier(n_neighbors=4)

# Fit the model on the training data
# This step trains the KNN classifier on the training dataset comprising of 'x_train' features and 'y_train' target labels.
knn.fit(x_train, y_train)
```

```python
# Predicting the model
y_pred = knn.predict(x_test)
```

EVALUATING

```python
from sklearn.metrics import accuracy_score, confusion_matrix , recall_score ,precision_score

# 1. Accuracy Score: 
a = accuracy_score(y_test, y_pred)
# 2. Recall Score
r = recall_score(y_test,y_pred)
# 3. Precision Score
p = precision_score(y_test,y_pred)

print("precision_score:" , p)
print("Recall_score:", r)
print("Accuracy Score: ", a)
print()

# 2. Confusion Matrix:
print("Confusion Matrix: ")
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True)
plt.show()
```
